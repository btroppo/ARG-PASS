```python
import os
import shutil
import pandas as pd

# Paths
pdb_source_dir = '/example/sub_class_B3/'
cluster_base_dir = '/example/sub_class_B3/cluster_20/'
cluster_file_path = '/example/sub_class_B3/20_cluster.tsv'  

# Load the cluster data
cluster_data = pd.read_csv(cluster_file_path, sep='\t', header=None, names=["Cluster", "Member"])

# Ensure the destination base directory exists
os.makedirs(cluster_base_dir, exist_ok=True)

# Iterate through the cluster data to organise files
for cluster, member in cluster_data.itertuples(index=False):
    # Create a directory for the cluster if it doesn't exist
    cluster_dir = os.path.join(cluster_base_dir, cluster)
    os.makedirs(cluster_dir, exist_ok=True)
    
    # Path to the member PDB file
    member_file = os.path.join(pdb_source_dir, f"{member.strip()}.pdb")
    
    # Move the file if it exists
    if os.path.isfile(member_file):
        shutil.copy(member_file, os.path.join(cluster_dir, os.path.basename(member_file)))
        #print(f"Moved: {member_file} -> {cluster_dir}")
    else:
        print(f"File not found: {member_file}")
```
